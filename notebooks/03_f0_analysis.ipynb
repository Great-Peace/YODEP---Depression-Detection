# 03 — F0 Analysis

F0 contour analysis (Extra 1), Wilcoxon tests (Extra 2), actor quality table (Extra 3).

```python
import sys
sys.path.insert(0, '..')

from pathlib import Path
import yaml

from src.data.yodep_loader import load_yodep_manifest
from src.evaluation.significance import test_f0_difference_per_speaker, summarise_significance
from src.visualisation.f0_contours import plot_all_speakers
from src.visualisation.results_tables import build_table5_wilcoxon_summary, save_table
from src.utils.logger import setup_logging

setup_logging()

with open('../config/config.yaml') as f:
    cfg = yaml.safe_load(f)

df = load_yodep_manifest(Path('../data/yodep/raw'))
print(f'Loaded {len(df)} recordings')
```

```python
# Extra 1: F0 contour plots for all speakers
figures_dir = Path('../results/figures')
plot_all_speakers(df, cfg['audio'], figures_dir)
print('F0 contour plots saved to', figures_dir)
```

```python
# Extra 2: Wilcoxon signed-rank test per speaker per language
sig_df = test_f0_difference_per_speaker(df, cfg['audio'])
print(sig_df.to_string())
```

```python
# Table 5: Wilcoxon summary
summary = summarise_significance(sig_df)
print(summary)

t5 = build_table5_wilcoxon_summary(summary)
save_table(t5, Path('../results/tables'), 'table5_wilcoxon',
           caption='Wilcoxon signed-rank test summary: F0 difference (NORMAL vs DEPRESSED) per language.',
           label='tab:wilcoxon')
```
